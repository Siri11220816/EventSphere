```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://pg.casel.org/review-programs/"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)

print("Website status:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")

programs = []

# Find links that point to individual CASEL program pages
program_links = soup.select('a[href^="https://pg.casel.org/"]')

seen = set()

for link in program_links:
    name = link.get_text(" ", strip=True)
    href = link.get("href", "").strip()

    # Keep only likely program-detail pages
    if (
        name
        and href
        and href != url
        and href.rstrip("/") != "https://pg.casel.org"
        and name not in seen
        and name not in [
            "View Full Details",
            "Visit CASEL.org",
            "Program Guide",
            "View the program list",
            "Explore all SEL programs"
        ]
    ):
        # Avoid navigation/filter links
        if "/review-programs/" not in href and "/connect-your-criteria" not in href:
            seen.add(name)
            programs.append({
                "program_name": name,
                "program_url": href
            })

# Remove obvious duplicates
program_df = pd.DataFrame(programs).drop_duplicates(
    subset=["program_name"]
).reset_index(drop=True)

# Save the collected program information
output_path = "data/raw/casel_programs.csv"

program_df.to_csv(output_path, index=False)

print("Collection attempt completed!")
print("Programs collected:", len(program_df))
print("Saved to:", output_path)

program_df.head(10)
```

```python
import pandas as pd

input_path = "data/raw/casel_programs.csv"
output_path = "data/processed/casel_programs_clean.csv"

program_df = pd.read_csv(input_path)

# Website/navigation pages that are not SEL programs
remove_names = [
    "Forgot your password?",
    "Identify Your Goals",
    "Compare Programs",
    "About the Program Guide",
    "FAQs",
    "For Providers",
    "Program Guide",
    "Visit CASEL.org"
]

clean_df = program_df[
    ~program_df["program_name"].isin(remove_names)
].copy()

# Remove the generic ".b" page
clean_df = clean_df[
    clean_df["program_name"].str.strip() != ".b"
]

# Remove duplicate program names
clean_df = clean_df.drop_duplicates(
    subset=["program_name"]
).reset_index(drop=True)

# Save cleaned dataset
clean_df.to_csv(output_path, index=False)

print("Cleaning completed!")
print("Original rows:", len(program_df))
print("Clean rows:", len(clean_df))
print("Saved to:", output_path)

print("\nFirst 10 programs:")
display(clean_df.head(10))
```

```python
import pandas as pd

# Load our SEL dataset
df = pd.read_csv("data/processed/sel_concepts.csv")

# Remove missing values
df = df.dropna()

# Clean text
df["competency"] = df["competency"].str.strip()
df["description"] = df["description"].str.strip()
df["skill"] = df["skill"].str.strip()

# Remove duplicate records
df = df.drop_duplicates()

# Create a searchable text field
df["knowledge_text"] = (
    df["competency"] + ". " +
    df["description"] + " Skill: " +
    df["skill"]
)

# Save preprocessed dataset
output_path = "data/processed/sel_concepts_preprocessed.csv"
df.to_csv(output_path, index=False)

print("Preprocessing completed!")
print("Rows:", len(df))
print("Columns:", list(df.columns))
print("Saved to:", output_path)

display(df)
```

```python
import pandas as pd

# Load preprocessed SEL knowledge
sel_df = pd.read_csv(
    "data/processed/sel_concepts_preprocessed.csv"
)

def sel_assistant(query):
    query = query.lower()

    # Find records containing words from the user's query
    matches = []

    for _, row in sel_df.iterrows():
        searchable_text = row["knowledge_text"].lower()

        if any(word in searchable_text for word in query.split()):
            matches.append(row)

    if len(matches) == 0:
        return """
I could not find a closely related SEL concept in my current
knowledge base.

Try asking about emotions, stress management, empathy,
communication, relationships, or decision-making.
"""

    # Use the first matching concept
    result = matches[0]

    response = f"""
SEL Assistant

Relevant SEL Competency:
{result['competency']}

Related Skill:
{result['skill']}

Explanation:
{result['description']}

This information is based on the CASEL SEL framework.
"""

    return response
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
# Create embeddings for all SEL knowledge records

sel_embeddings = embedding_model.encode(
    sel_df["knowledge_text"].tolist(),
    normalize_embeddings=True
)

print("SEL embeddings created successfully!")
print("Number of records:", len(sel_embeddings))
print("Embedding size:", len(sel_embeddings[0]))
```

```python
import numpy as np

def retrieve_sel(query, top_k=3):
    # Convert user question into an embedding
    query_embedding = embedding_model.encode(
        [query],
        normalize_embeddings=True
    )[0]

    # Calculate similarity with all SEL records
    scores = np.dot(sel_embeddings, query_embedding)

    # Get the highest-scoring records
    top_indices = np.argsort(scores)[::-1][:top_k]

    results = sel_df.iloc[top_indices].copy()
    results["similarity_score"] = scores[top_indices]

    return results
```

```python
results = retrieve_sel("How can I manage my emotions?", top_k=3)

results[[
    "competency",
    "skill",
    "similarity_score"
]]
```

```python
test_questions = [
    "How can I manage my emotions?",
    "How can I understand other people's feelings?",
    "How can I improve communication?",
    "How can I solve problems responsibly?"
]

for question in test_questions:
    results = retrieve_sel(question, top_k=1)

    result = results.iloc[0]

    print("=" * 60)
    print("QUESTION:", question)
    print("COMPETENCY:", result["competency"])
    print("SKILL:", result["skill"])
    print("SIMILARITY:", round(result["similarity_score"], 3))
```
